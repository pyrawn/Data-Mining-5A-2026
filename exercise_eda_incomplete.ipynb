```python
# Install ucimlrepo if needed
try:
    from ucimlrepo import fetch_ucirepo
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'ucimlrepo', '-q'])
    from ucimlrepo import fetch_ucirepo

repo = fetch_ucirepo(id=560)
df = pd.concat([repo.data.features, repo.data.targets], axis=1)
print(f'Dataset loaded — shape: {df.shape}')
```

```python
# ── Preprocessing ──────────────────────────

# Drop Date column
if 'Date' in df.columns:
    df.drop(columns=['Date'], inplace=True)
    print('Dropped: Date')

# Encode Functioning Day: Yes -> 1, No -> 0
df['Functioning Day'] = df['Functioning Day'].map({'Yes': 1, 'No': 0})
print('Encoded: Functioning Day')

# The UCI loader sometimes returns extra columns named 'Temperature(°C)' and
# 'Wind speed (m/s)' that are entirely NaN (ghost duplicates of the real columns
# 'Temperature' and 'Wind speed'). Drop them before injecting artificial NaNs.
df.drop(columns=['Temperature(°C)', 'Wind speed (m/s)'], errors='ignore', inplace=True)
print('Dropped ghost columns (if present): Temperature(°C), Wind speed (m/s)')

# Introduce artificial missing values to simulate IoT sensor dropout
# Target columns are the actual working columns: 'Temperature' and 'Wind speed'
rng = np.random.default_rng(RANDOM_SEED)
temp_idx = rng.choice(df.index, size=int(0.04 * len(df)), replace=False)
wind_idx = rng.choice(df.index, size=int(0.03 * len(df)), replace=False)
df.loc[temp_idx, 'Temperature'] = np.nan
df.loc[wind_idx, 'Wind speed']  = np.nan
print(f'Missing values introduced — Temperature: {df["Temperature"].isna().sum()} '
      f'({100*df["Temperature"].isna().mean():.1f}%), '
      f'Wind speed: {df["Wind speed"].isna().sum()} '
      f'({100*df["Wind speed"].isna().mean():.1f}%)')
```

### What YData Profiling automates — and what it cannot replace

**What it automates:**
- Per-variable type inference, summary statistics, missing value counts, and cardinality checks.
- Correlation matrices (Pearson, Spearman, Cramér's V) across all detected variable types.
- Distribution histograms for every column, interaction plots, and a complete HTML report in a single call.
- Duplicate row detection and constant/quasi-constant column flags.

**What it cannot replace:**
- **Domain-driven preprocessing decisions** — encoding ordinal variables correctly, choosing whether to impute or drop missing values, or deciding which outlier treatment is appropriate require subject-matter expertise.
- **Tailored visualizations** — the automated charts are generic; a bespoke scatter matrix or custom color-coded chart communicates insights far more clearly to a specific audience.
- **Hypothesis-driven exploration** — profiling tells you *what* is in the data; it does not help you ask *why* a variable behaves a certain way or *how* to engineer meaningful features.
- **Contextual annotation** — every insight in this notebook was paired with an explanation tied to the problem domain (bike sharing demand). Automated tools produce statistics without narrative.

**Why we built everything manually first:**  
Working through each section step-by-step forces you to understand *why* each metric matters, *how* to compute it, and *what decisions* it informs. Using the profiling tool as a final cross-check — rather than a starting point — reinforces these skills and prevents treating EDA as a black box.
